1. summary(cars)
이 코드는 기본적으로 R에서 제공하는 함수 호출 방식입니다. summary() 함수는 데이터를 요약하는 함수로, 데이터 프레임이나 벡터 등의 요약 통계를 보여줍니다.
여기서는 cars 데이터셋에 직접 summary() 함수를 호출하여 cars 데이터셋에 대한 요약 정보를 제공합니다.
r
코드 복사
summary(cars)
이 방식에서는 함수가 데이터를 직접 받아서 처리합니다.
결과는 cars 데이터셋의 요약 통계가 출력됩니다.
2. cars %>% summary()
이 코드는 파이프 연산자 %>%를 사용하여 데이터를 함수를 통해 전달합니다. 파이프 연산자의 역할은 왼쪽에 있는 데이터를 오른쪽 함수의 첫 번째 인자로 전달하는 것입니다.
즉, cars %>% summary()는 cars 데이터를 먼저 선택하고, 그 데이터를 summary() 함수에 넘겨서 요약 정보를 출력합니다.
r
코드 복사
cars %>% summary()
이 방식은 여러 함수가 연달아 있을 때 특히 유용합니다. 예를 들어 데이터를 전처리하고, 요약하거나 분석하는 과정에서 데이터 흐름을 더 직관적으로 표현할 수 있습니다.
차이점:
기능적으로는 두 코드는 동일합니다. 하지만 가독성과 코드 작성 방식에서 차이가 있습니다.
summary(cars)는 함수가 데이터를 직접 처리하는 전통적인 방식이고, cars %>% summary()는 파이프 연산자를 사용하여 데이터가 함수로 흐르도록 표현한 것입니다. 후자는 특히 여러 함수가 연결되는 복잡한 코드에서 훨씬 더 읽기 쉽습니다.
